```python
import os
from openai import OpenAI, AuthenticationError, APIConnectionError, APITimeoutError
from dotenv import load_dotenv

# Load environment variables from .env file
load_dotenv()

# Initialize the OpenAI client
api_key = os.getenv('OPENAI_API_KEY', '').strip().strip('"').strip("'")
if not api_key:
    raise ValueError('OPENAI_API_KEY not found. Please set it in your .env file')

if not api_key.startswith(('sk-', 'sk-proj-')):
    print('Warning: OPENAI_API_KEY format looks unusual. Double-check your .env value.')

proxy = os.getenv('HTTPS_PROXY') or os.getenv('HTTP_PROXY')
if proxy:
    print('Proxy detected from environment variables.')

# Add sensible timeout/retry defaults and avoid compressed responses if local httpx2/zstd is incompatible.
client = OpenAI(
    api_key=api_key,
    timeout=30.0,
    max_retries=2,
    default_headers={'Accept-Encoding': 'identity'},
)
print('✓ OpenAI client initialized successfully!')
```

```python
from pathlib import Path
from threading import Thread
from werkzeug.serving import make_server
from IPython.display import HTML, display
from app import app

class NotebookServer:
    def __init__(self, flask_app, host='127.0.0.1', port=5000):
        self.server = make_server(host, port, flask_app, threaded=True)
        self.thread = Thread(target=self.server.serve_forever, daemon=True)
    def start(self):
        self.thread.start()
    def stop(self):
        self.server.shutdown()

if 'meal_planner_server' in globals():
    meal_planner_server.stop()
meal_planner_server = NotebookServer(app)
meal_planner_server.start()
display(HTML('<h3>Meal Planner is running</h3><p><a href="http://127.0.0.1:5000" target="_blank">Open the Weekly Meal Planner</a></p><p>To stop the server, run <code>meal_planner_server.stop()</code>.</p>'))
```
